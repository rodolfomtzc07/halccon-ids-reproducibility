```python
cat_cols = ['proto', 'service', 'state', 'attack_cat']

rows = []
for col in cat_cols:
    train_uniques = train_df[col].astype(str).nunique(dropna=False)
    test_uniques = test_df[col].astype(str).nunique(dropna=False)

    train_set = set(train_df[col].astype(str).unique())
    test_set = set(test_df[col].astype(str).unique())

    only_train = sorted(train_set - test_set)
    only_test = sorted(test_set - train_set)
    common = sorted(train_set & test_set)

    rows.append({
        'column': col,
        'train_nunique': train_uniques,
        'test_nunique': test_uniques,
        'common_count': len(common),
        'only_train_count': len(only_train),
        'only_test_count': len(only_test)
    })

cardinality_df = pd.DataFrame(rows).sort_values('train_nunique', ascending=False)
print(cardinality_df)

for col in cat_cols:
    print(f"\n=== {col} ===")
    print("TRAIN nunique:", train_df[col].astype(str).nunique(dropna=False))
    print("TEST  nunique:", test_df[col].astype(str).nunique(dropna=False))

    train_set = set(train_df[col].astype(str).unique())
    test_set = set(test_df[col].astype(str).unique())

    print("Only in TRAIN:", sorted(train_set - test_set))
    print("Only in TEST :", sorted(test_set - train_set))

    print("\nTop TRAIN categories:")
    print(train_df[col].astype(str).value_counts(dropna=False).head(15))

    print("\nTop TEST categories:")
    print(test_df[col].astype(str).value_counts(dropna=False).head(15))
```

```python
from sklearn.preprocessing import LabelEncoder

# =========================
# 1) Definir target y drops
# =========================
target_col = 'attack_cat'
drop_cols = ['id', 'label', 'attack_cat']   # para X en multiclase

# =========================
# 2) Separar X e y
# =========================
X_train = train_df.drop(columns=drop_cols).copy()
X_test  = test_df.drop(columns=drop_cols).copy()

y_train_raw = train_df[target_col].copy()
y_test_raw  = test_df[target_col].copy()

# =========================
# 3) Label encoding del target
# =========================
label_encoder = LabelEncoder()

y_train = label_encoder.fit_transform(y_train_raw)
y_test  = label_encoder.transform(y_test_raw)

# =========================
# 4) Verificaciones
# =========================
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)

print("\nColumnas de X_train:")
print(X_train.columns.tolist())

print("\nColumnas de X_test:")
print(X_test.columns.tolist())

print("\n¿Mismas columnas en X_train y X_test?")
print(list(X_train.columns) == list(X_test.columns))

print("\nClases del target codificadas:")
for i, cls in enumerate(label_encoder.classes_):
    print(f"{i} -> {cls}")
```

```python
from category_encoders import CatBoostEncoder

# =========================================
# 1) Identificar columnas categóricas en X
# =========================================
cat_cols = ['proto', 'service', 'state']

print("Categorical columns:", cat_cols)

# =========================================
# 2) Copias de trabajo
# =========================================
X_train_cb = X_train.copy()
X_test_cb = X_test.copy()

# Asegurar tipo string en categóricas
for col in cat_cols:
    X_train_cb[col] = X_train_cb[col].astype(str)
    X_test_cb[col] = X_test_cb[col].astype(str)

# =========================================
# 3) CatBoost Encoding
# =========================================
cb_encoder = CatBoostEncoder(cols=cat_cols)

# fit SOLO en train
X_train_cb[cat_cols] = cb_encoder.fit_transform(X_train_cb[cat_cols], y_train)

# transform en test
X_test_cb[cat_cols] = cb_encoder.transform(X_test_cb[cat_cols])

# =========================================
# 4) Verificaciones
# =========================================
print("X_train_cb shape:", X_train_cb.shape)
print("X_test_cb shape :", X_test_cb.shape)

print("\nDtypes after CatBoost encoding:")
print(X_train_cb.dtypes)

print("\nPrimeras filas de X_train_cb:")
print(X_train_cb.head())

print("\nPrimeras filas de X_test_cb:")
print(X_test_cb.head())
```
